### Now with the actual analysis
- 2010 - 2024
- All banks and quarterly financial records

```python
all_batches = []
offset = 0

while True:
    url = f"https://api.fdic.gov/banks/financials?fields=CERT,REPDTE,NAME,ASSET,DEP,ROA,ROE,NETINC&filters=REPDTE:[2010-01-01 TO 2025-12-31]&limit=1000&offset={offset}"
    response = requests.get(url)
    data = response.json()

    batch_df = pd.DataFrame([record["data"] for record in data["data"]])
    if batch_df.empty:
        break

    all_batches.append(batch_df)
    offset += 1000

data_df = pd.concat(all_batches, ignore_index=True)
```

```python
data_df.to_csv("../data/fdic_financials.csv", index=False)
```

### Average ROA

```python
print("Min ROA:", data_df["ROA"].min())
print("Max ROA:", data_df["ROA"].max())
average_ROA = data_df["ROA"].mean().round(7)
print("Average ROA:", average_ROA)
```

 ### Highest and Lowest 5 ROA's

```python
print(data_df.nlargest(5, "ROA")[["NAME", "REPDTE","ASSET","ROA"]])
print(data_df.nsmallest(5, "ROA")[["NAME", "REPDTE","ASSET","ROA"]])
```

### Negative ROA

```python
negative_ROA = data_df[data_df["ROA"] < 0]
print("Num of records with negative ROA:", len(negative_ROA))
```

```python
median = data_df["ROA"].median()
print("Median ROA:", median.round(7))

perct_25 = data_df["ROA"].quantile(0.25)
print("25th Percentile:", perct_25.round(7))

perct_75 = data_df["ROA"].quantile(0.75)
print("75th Percentile:", perct_75.round(7))

print("Difference between mean and median ROA's:", (average_ROA - median).round(7))
```
