```python
import requests
import pandas as pd

url = "https://api.fdic.gov/banks/financials?limit=10" ##ONLY 10 FOR NOW

response = requests.get(url)
print("Status Code:", response.status_code)
data = response.json()
print(data.keys())
```

```python
data_df = pd.DataFrame(data['data'])
data_df
```

- data needs splitting

```python
print(data["data"][0]) 

##ONE ROW OF "data" column have all of the information -> split
```

```python
data_df = pd.DataFrame([record["data"] for record in data["data"]])
data_df.head(3)
```

- correctly splitted the data column into multiple ones and created a better dataframe
- One row: one bank's financial information for a specific reporting period (usually a quarter).
